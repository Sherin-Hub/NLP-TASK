```python
import pandas as pd
import re
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
df=pd.read_csv("IMDB Dataset.csv")
print("Dataset loaded successfully!")
print("Shape of dataset:", df.shape)
```

```python
# ============================================================
# 4. DISPLAY FIRST 5 REVIEWS
# ============================================================

print("\nFirst 5 rows:")
display(df.head())


# ============================================================
# 5. CHECK DATASET INFORMATION
# ============================================================

print("\nColumn names:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum())


# ============================================================
# 6. REMOVE MISSING REVIEWS
# ============================================================

df = df.dropna(subset=['review'])

print("\nMissing values after removal:")
print(df.isnull().sum())


# ============================================================
# 7. CHECK DUPLICATE REVIEWS
# ============================================================

duplicates = df['review'].duplicated().sum()

print("\nNumber of duplicate reviews:", duplicates)


# ============================================================
# 8. REMOVE DUPLICATE REVIEWS
# ============================================================

df = df.drop_duplicates(subset=['review'])

print("Dataset shape after removing duplicates:", df.shape)

stop_words=set(stopwords.words('english'))
lemmatizer=WordNetLemmatizer()
def clean_text(text):
    text=text.lower()
    text=re.sub(r'<.*?>','',text)
    text=re.sub(r'http\S+|www\S+','',text)
    text=re.sub(r'\d+','',text)
    text=re.sub(r'[^A-Za-z\s]','',text)
    tokens=word_tokenize(text)
    token=[word for word in tokens 
           if word not in stop_words
    ]
    tokens=[lemmatizer.lemmatize(word)for word in tokens]
    return ' '.join(tokens)
    # ============================================================
# 11. TEST THE CLEANING FUNCTION
# ============================================================

example = "This movie was AMAZING!!!"

print("\nOriginal example:")
print(example)

print("\nCleaned example:")
print(clean_text(example))


# ============================================================
# 12. APPLY CLEANING TO ALL REVIEWS
# ============================================================

print("\nCleaning all reviews...")

df['cleaned_review'] = df['review'].apply(clean_text)

print("Cleaning completed!")


# ============================================================
# 13. COMPARE ORIGINAL AND CLEANED REVIEWS
# ============================================================

print("\nOriginal vs Cleaned Reviews:")

display(
    df[['review', 'cleaned_review']].head(10)
)


# ============================================================
# 14. DISPLAY FINAL DATASET
# ============================================================

print("\nFinal dataset:")
display(df.head())


# ============================================================
# 15. CHECK FINAL COLUMNS
# ============================================================

print("\nFinal columns:")
print(df.columns.tolist())


# ============================================================
# 16. EXPORT CLEANED DATASET AS CSV
# ============================================================

output_file = "IMDB_Cleaned_Dataset.csv"

df.to_csv(output_file, index=False)

print("\nDataset exported successfully!")
print("File name:", output_file)


# ============================================================
# 17. FINAL SUMMARY
# ============================================================

print("\n========================================")
print("       PREPROCESSING COMPLETED")
print("========================================")
print("Total reviews:", len(df))
print("Columns:", df.columns.tolist())
print("Output file:", output_file)
```
